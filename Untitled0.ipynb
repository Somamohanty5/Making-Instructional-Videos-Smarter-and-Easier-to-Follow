# Load the datasetDataset: PubMed RCT 20k

https://huggingface.co/datasets/armanc/pubmed-rct20k

Task:
Classify sentences in biomedical abstracts by their rhetorical role:
Background, Objective, Methods, Results, or Conclusion.

# Loading Dataset

```python
from datasets import load_dataset
import pandas as pd

# Loading PubMed RCT 20k
dataset = load_dataset("armanc/pubmed-rct20k")

# Converting  to pandas
train_df = pd.DataFrame(dataset['train'])
val_df   = pd.DataFrame(dataset['validation'])
test_df  = pd.DataFrame(dataset['test'])

print("Train shape:", train_df.shape)
train_df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
train_df['label'] = le.fit_transform(train_df['label'])
val_df['label']   = le.transform(val_df['label'])
test_df['label']  = le.transform(test_df['label'])

print("Labels:", list(le.classes_))
```

# MODEL 1: TF-IDF + SVM

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report

# Vectorize text
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
X_train = vectorizer.fit_transform(train_df['text'])
X_val   = vectorizer.transform(val_df['text'])
X_test  = vectorizer.transform(test_df['text'])

# Train
svm_model = LinearSVC()
svm_model.fit(X_train, train_df['label'])

y_pred_svm = svm_model.predict(X_test)
print("TF-IDF + SVM Results:")
print(classification_report(test_df['label'], y_pred_svm, target_names=le.classes_))
```

# MODEL 2: Word2Vec + Logistic Regression




```python
from gensim.models import Word2Vec
import numpy as np
from sklearn.linear_model import LogisticRegression

# Tokenize sentences
sentences = [t.split() for t in train_df['text']]
w2v = Word2Vec(sentences, vector_size=100, window=5, min_count=2, workers=4)

# Converting each sentence to mean vector
def vectorize_sentence(s, model):
    words = [w for w in s.split() if w in model.wv]
    if not words:
        return np.zeros(model.vector_size)
    return np.mean(model.wv[words], axis=0)

X_train_w2v = np.array([vectorize_sentence(t, w2v) for t in train_df['text']])
X_test_w2v  = np.array([vectorize_sentence(t, w2v) for t in test_df['text']])


lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_w2v, train_df['label'])

y_pred_w2v = lr.predict(X_test_w2v)
print("Word2Vec + Logistic Regression Results:")
print(classification_report(test_df['label'], y_pred_w2v, target_names=le.classes_))
```

# MODEL 3: BERT Fine Tuning

```python
from datasets import Dataset
from transformers import (
    AutoTokenizer, AutoModelForSequenceClassification,
    Trainer, TrainingArguments, EarlyStoppingCallback
)
import numpy as np
import pandas as pd
import evaluate
from sklearn.metrics import classification_report

#Label Mapping
label_names = ["background", "conclusions", "methods", "objective", "results"]
label2id = {name: idx for idx, name in enumerate(label_names)}
id2label = {idx: name for name, idx in label2id.items()}

print("Label Mapping:")
print(label2id)

train_dataset = Dataset.from_pandas(train_df[["text", "label"]])
test_dataset  = Dataset.from_pandas(test_df[["text", "label"]])


def finetune_eval(model_name, output_dir):
    print(f"\n Fine-tuning {model_name} ")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSequenceClassification.from_pretrained(
        model_name,
        num_labels=len(label2id),
        id2label=id2label,
        label2id=label2id
    )

    # Tokenization
    def tok_fn(batch):
        return tokenizer(batch["text"], truncation=True, padding="max_length", max_length=128)
    tr_tok = train_dataset.map(tok_fn, batched=True)
    te_tok = test_dataset.map(tok_fn, batched=True)
    tr_tok.set_format("torch", columns=["input_ids", "attention_mask", "label"])
    te_tok.set_format("torch", columns=["input_ids", "attention_mask", "label"])

    #  Metrics
    metric_acc = evaluate.load("accuracy")
    metric_f1 = evaluate.load("f1")

    def metrics(p):
        preds = np.argmax(p.predictions, axis=-1)
        return {
            "accuracy": metric_acc.compute(predictions=preds, references=p.label_ids)["accuracy"],
            "f1": metric_f1.compute(predictions=preds, references=p.label_ids, average="weighted")["f1"]
        }

    # Training arguments
    args = TrainingArguments(
        output_dir=output_dir,
        evaluation_strategy="epoch",
        save_strategy="epoch",
        num_train_epochs=3,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=16,
        learning_rate=2e-5,
        weight_decay=0.01,
        logging_steps=500,
        load_best_model_at_end=True,
        metric_for_best_model="f1",
        report_to=[],
        fp16=True,
        dataloader_num_workers=2,
        max_steps=4000
    )

    # Trainer setup
    trainer = Trainer(
        model=model,
        args=args,
        train_dataset=tr_tok,
        eval_dataset=te_tok,
        tokenizer=tokenizer,
        compute_metrics=metrics,
        callbacks=[EarlyStoppingCallback(early_stopping_patience=1)]
    )

    trainer.train()
    res = trainer.evaluate()

    print(f"\n {model_name} Evaluation Results:")
    print(f"Accuracy: {res['eval_accuracy']:.4f}, F1: {res['eval_f1']:.4f}, Loss: {res['eval_loss']:.4f}")

    # Classification Report
    pred_output = trainer.predict(te_tok)
    preds = np.argmax(pred_output.predictions, axis=-1)
    print("\n Classification Report:")
    print(classification_report(
        pred_output.label_ids,
        preds,
        target_names=list(label2id.keys()),
        digits=2
    ))

    # Save CSV Report
    report = classification_report(
        pred_output.label_ids,
        preds,
        target_names=list(label2id.keys()),
        output_dict=True
    )
    pd.DataFrame(report).transpose().to_csv(f"{model_name.replace('/', '_')}_classification_report.csv", index=True)

    return res


# Running bith models

res_bert_base = finetune_eval("bert-base-uncased", "./bert_base_fast")
res_pubmed    = finetune_eval("microsoft/BiomedNLP-PubMedBERT-base-uncased-abstract", "./pubmed_fast")


summary = pd.DataFrame([
    {"Model": "BERT-base", "Accuracy": res_bert_base["eval_accuracy"], "F1": res_bert_base["eval_f1"], "Loss": res_bert_base["eval_loss"]},
    {"Model": "PubMedBERT", "Accuracy": res_pubmed["eval_accuracy"], "F1": res_pubmed["eval_f1"], "Loss": res_pubmed["eval_loss"]}
])

print("\n Final Comparison Summary:")
print(summary)
```
